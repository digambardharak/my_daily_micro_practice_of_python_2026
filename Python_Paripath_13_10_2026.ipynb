1. Write a Python program using a for loop to print numbers from 1 to 5.

```python
for i in range(1,6):
    print(i)
```

2. Write a Python program using a while loop to print even numbers from 2 to 10.

```python
number = 2
while number <= 10:
    if number % 2 == 0:
        print(number)
        number = number + 2
```

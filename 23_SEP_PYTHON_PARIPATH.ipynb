<span style="color: red;">**For loop**</span>

**Q1. Write a Python program using a `for` loop to print numbers from 1 to 10.**


```python
for i in range(1,11):
    print(i)
```

**Q2. Write a Python program using a `for` loop to print each element of the list `[10, 20, 30, 40, 50]`.**

```python
list1 = [10,20,30,40,50]
for i in list1:
    print(i)
```

**Q1. Write a Python program using a `while` loop to print numbers from 1 to 10.**



```python
count = 1
while count <= 10:
    print(count)
    count += 1
```

**Q2. Write a Python program using a `while` loop to print even numbers from 2 to 10.**

```python
even = 2
while even <= 10:
    print(even)
    even += 2
```

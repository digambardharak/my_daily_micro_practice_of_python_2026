Q1. Write a Python program using a while loop to print this pattern:

*
**
***

```python
for i in range(1,4):
    for j in range(i):
        print("*", end = " ")
    print()
 
```

Write a Python program using a while loop to print this number pattern:


1
12
123

```python
for i in range(1, 4):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

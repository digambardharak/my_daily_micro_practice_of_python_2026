**1. Write a Python program to find the largest number in the list `[10, 25, 7, 42, 18]` without using `max()`.**

```python
numbers = [10,25,7,42,18]

#step1 Assume the firse element is largest
largest = numbers[0]

#step 2 check every numbers
for i in numbers:

    #step 3 compare current number with largest
    if i > largest:

        #step 4 update largest
        largest = i
print(largest)
```

**2. Write a Python program to count how many even numbers are present in the list `[12, 7, 4, 19, 8, 15]`.**

```python
numbers = [12,7,4,19,8,15]

#step 1 for loop
for i in numbers:

    #step 2 check even or odd
    if i % 2 == 0:

        print(f"{i} is even")

    else:
        print(f"{i} is odd")
```

```python
numbers = [12,7,4,19,8,15]

count = 0

for i in numbers:
    if i % 2 == 0:
        count = count + 1

print(count)
```

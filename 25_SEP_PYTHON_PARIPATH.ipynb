**Error Note — TypeError: 'int' object is not iterable**

Mistake: I used for i in num, but num is an integer (10), and an integer cannot be iterated directly.
How to solve: Create a sequence using range(), e.g. for i in range(1, num + 1):, and use total = total + i to add each number.
Remember: Before using for i in X, ask “Is X iterable?”; if I need numbers from 1 to num, use range(1, num + 1).

```python
def sum_odd_even(num):
    total = 1
    for i in range(1, num+1):
        total = total+1

    return total
result = sum_odd_even(10)
print(result)
if result%2==0:
    print("even")
else:
    print("odd")
```

**Error Note — Sum vs Counting**

Mistake: I used total = 1 and total = total + 1, which counts loop iterations instead of calculating the sum.
How to solve: Start with total = 0 and add the current loop value using total = total + i.
Remember: Counting → +1, Summing → +i; for a sum, initialize total = 0.

```python
def sum_odd_even(num):
    total = 0
    for i in range(1, num + 1):
        total = total + 1
    return total
result = sum_odd_even(10)
print(result)
if result % 2 == 0:
    print("even")
else:
    print("odd")
```

**Mistake Note — total + 1 vs total + i**

Mistake: Used total = total + 1, which counts loop iterations instead of adding the actual numbers.
Correction: Use total = total + i because i is the current number being added.
Remember: Counting → +1 | Summing → +i.

```python
def sum_odd_even(num): # define function function name is sum_odd_even
    total = 0           # i initialize total is zero
    for i in range(1, num +1):  # in t=for loop i give range from 1 to this given num atributes.
        total = total + i      # that total increasing by ading current value of i as per itreation
    return total            # this returens total value

result = sum_odd_even(10)  # call the function
print(result)
if result%2==0:     # check result is even or odd
    print("even")
else:
    print("odd")
```

**Difference between this aboved and belowed codes**
Difference: range(1, num) stops before num, while range(1, num + 1) includes num.

Example: If num = 10, range(1, num) generates 1–9 → Sum = 45, while range(1, num + 1) generates 1–10 → Sum = 55 .

Remember: range() excludes the ending value, so to include num, use range(1, num + 1).

```python
def sum_odd_even(num):
    total = 0
    for i in range(1, num ):
        total = total + i
    return total

result = sum_odd_even(10)
print(result)
if result%2==0:
    print("even")
else:
    print("odd")
```

**2. Write a Python function that takes two numbers as arguments, uses operators to calculate their sum, and uses a conditional statement and loop to print the result multiple times based on the given values.**

```python
def multy_result_sum(num1, num2):
    total = num1 + num2

    if num1 > 0:
        for i in range(num2):
            print(total)
    else:
        print("This is a positive ")

result = multy_result_sum(10, 20)
print(result)
```

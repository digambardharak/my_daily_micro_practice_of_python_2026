**1. Write a Python function that takes two numbers as arguments and returns their sum.**

```python
def add_num(num1,num2):  # we create a add_num function 
    return num1 + num2   # in return we perform addition operation

result = add_num(10, 20)
print(result)
```

**2. Write a Python function that takes a number as an argument and returns whether it is even or odd.**

```python
def check_odd_even(number):
    if number%2==0:
        print(f"{number} is even")
    else:
        print(f"{number} is odd")

result = check_odd_even(4)
 
```

```python
def even_or_odd(number):
    if number%2==0:
        return "even"
    else:
        return "odd"

result = even_or_odd(5)
print(result)
```

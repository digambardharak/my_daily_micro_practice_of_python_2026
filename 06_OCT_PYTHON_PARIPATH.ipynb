**1. Write a Python program to sort the list `[45, 12, 78, 23, 9]` in ascending order without using `sort()` or `sorted()`.**


```python
numbers = [2,4,5,3,1]

for i in range(len(numbers)):
    for j in range(i+1, len(numbers)):
        if numbers[i] > numbers[j]:
            numbers[i], numbers[j] = numbers[j], numbers[i]
print(numbers)
```

**2. Write a Python program to sort the list `[10, 5, 30, 15, 20]` in descending order without using `sort()` or `sorted()`.**

```python
numbers = [2,4,5,3,1]

for i in range(len(numbers)):
    for j in range(i+1, len(numbers)):
        if numbers[i] < numbers[j]:
            numbers[i], numbers[j] = numbers[j], numbers[i]
print(numbers)
```

**1. Write a Python program using nested `for` loops to print this pattern:**

```text
*
**
***
****
*****

````python
for i in range(5):
    for j in range(5):
        print("*")
````

````python
for i in range(5):
    
    for j in range(i):
        print("*",end="")
````

````python
for i in range(6):

    for j in range(i):
        print("*", end="")

    print()
````

**2. Write a Python program using nested `for` loops to print this pattern:**

```text
1
12
123
1234
```

````python
for i in range(1, 5):

    for j in range(i):
        print(i, end="")

    print()
````

````python
for i in range(1,5):

    for j in range(i):

        print(j, end="")
    print()
````

````python
for i in range(1,5):

    for j in range(1, i + 1):

        print(j, end="")
    print()
````

````python
# Outer loop controls the number of rows: 1 to 4
for i in range(1, 5):

    # Inner loop prints numbers from 1 up to the current row number
    for j in range(1, i + 1):

        # Print the current number on the same line
        print(j, end="")

    # Move to the next line after completing each row
    print()
````
